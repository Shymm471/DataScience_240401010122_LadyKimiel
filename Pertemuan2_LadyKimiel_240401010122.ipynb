```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print(df.head())
print(df.info())
```

## Eksplorasi Awal

- Jumlah baris dan kolom: (lihat output)
- Nama kolom: (lihat output)
- Missing values terdapat pada kolom Age, Cabin, dan Embarked.

```python
wk1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

print("Jumlah wanita kelas 1:")
print(len(wk1))

print("Jumlah yang selamat:")
print(wk1['Survived'].sum())

print("Rata-rata usia:")
print(wk1['Age'].mean())
```

```python
print(df.groupby('Pclass')['Survived'].mean())
```
