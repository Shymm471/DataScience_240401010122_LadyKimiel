## Tujuan

Mempelajari konsep Ensemble Learning menggunakan algoritma Random Forest untuk klasifikasi, memahami penanganan imbalanced dataset menggunakan class_weight, serta mengevaluasi model menggunakan Confusion Matrix, Accuracy, Precision, Recall, F1-Score, dan ROC-AUC.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
# Load dataset
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

# Melihat ukuran dataset
print("Shape :", df.shape)

# Menampilkan 5 data pertama
print("\n5 Data Pertama")
print(df.head())

# Melihat informasi dataset
print("\nInformasi Dataset")
print(df.info())

# Melihat distribusi target
print("\nDistribusi Target (Churn)")
print(df["Churn"].value_counts())

print("\nProporsi Target")
print(df["Churn"].value_counts(normalize=True))
```

```python
# Menghapus kolom customerID karena bukan fitur
df = df.drop("customerID", axis=1)

# Mengubah TotalCharges menjadi numerik
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Menghapus data yang kosong
df = df.dropna()

# Mengubah target Churn menjadi angka
df["Churn"] = df["Churn"].map({"No": 0, "Yes": 1})

# One-Hot Encoding untuk fitur kategorikal
df = pd.get_dummies(df, drop_first=True)

# Memisahkan fitur dan target
X = df.drop("Churn", axis=1)
y = df["Churn"]

# Membagi data train dan test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Data Train :", X_train.shape)
print("Data Test  :", X_test.shape)
```

```python
# Membangun model Random Forest
rf = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42
)

# Melatih model
rf.fit(X_train, y_train)

print("Model Random Forest berhasil dilatih!")
```

```python
# Prediksi
y_pred = rf.predict(X_test)
y_proba = rf.predict_proba(X_test)[:, 1]

# Evaluasi
print("=== Random Forest ===")
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print(f"\nAccuracy : {accuracy_score(y_test, y_pred):.3f}")
print(f"Precision: {precision_score(y_test, y_pred):.3f}")
print(f"Recall   : {recall_score(y_test, y_pred):.3f}")
print(f"F1-Score : {f1_score(y_test, y_pred):.3f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_proba):.3f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Probabilitas pelanggan mengalami churn
proba = rf.predict_proba(X_test)[:, 1]

hasil = pd.DataFrame({
    "Probabilitas Churn": proba
})

print("10 Probabilitas Churn Pertama")
print(hasil.head(10))
```

## Kesimpulan

Berdasarkan hasil pengujian, algoritma Random Forest berhasil digunakan untuk memprediksi customer churn dengan performa yang baik. Model mampu mengklasifikasikan data pelanggan menggunakan pendekatan ensemble learning serta menghasilkan nilai evaluasi yang tinggi. Selain itu, probabilitas churn yang dihasilkan melalui `predict_proba()` dapat dimanfaatkan untuk mengidentifikasi pelanggan yang berisiko berhenti berlangganan. Dengan demikian, Random Forest menjadi salah satu metode yang efektif untuk permasalahan klasifikasi pada dataset customer churn.
